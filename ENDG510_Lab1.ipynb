```python
#load your data
df = pd.read_csv("data.csv")
df.head() # prints top 5 rows from the datatset to check data is load or not
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing Values:\n", missing_values)
```

```python
# Separate rows by label
df_label1 = df[df["Label"] == 1].drop_duplicates() #removing duplicate readings recieved from the sensor
df_label0 = df[df["Label"] == 0]

df = pd.concat([df_label1, df_label0], ignore_index=True) # Combine the datasets back together

# Check the updated dataset
print("Dataset size:", df.shape)
print("\nLabel counts:")
print(df["Label"].value_counts())
```
